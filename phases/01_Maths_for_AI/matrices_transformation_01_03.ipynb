1) Apply rotation, scaling, and shearing to a unit square (corners at [0,0], [1,0], [1,1], [0,1]). Print the transformed corners for each. Verify that rotation preserves distances between corners.

```python
import numpy as np
square = np.array([
    [0, 0],
    [1, 0],
    [1, 1],
    [0, 1]
])

angle = np.radians(45)

R = np.array([
    [np.cos(angle), -np.sin(angle)],
    [np.sin(angle),  np.cos(angle)]
])

rotated = square @ R.T

# Scaling
S = np.array([
    [2, 0],
    [0, 3]
])

scaled = square @ S.T

# Shearing
H = np.array([
    [1, 1],
    [0, 1]
])

sheared = square @ H.T

print("Original:")
print(square)

print("\nRotated:")
print(rotated)

print("\nScaled:")
print(scaled)

print("\nSheared:")
print(sheared)

# Check distance between first two corners
original_distance = np.linalg.norm(square[0] - square[1])
rotated_distance = np.linalg.norm(rotated[0] - rotated[1])

print("\nOriginal distance:", original_distance)
print("Rotated distance:", rotated_distance)
print("Distance preserved:", np.isclose(original_distance, rotated_distance))
```

2) Find the eigenvalues of the matrix [[4, 2], [1, 3]] by hand using the characteristic equation. Then verify with your from-scratch function and with NumPy.

```python
import numpy as np

A = np.array([
    [4, 2], 
    [1, 3]
    ])

# 2. Compute eigenvalues and eigenvectors
eigenvalues, eigenvectors = np.linalg.eig(A)

print("Eigenvalues:")
print(eigenvalues)

print("\nEigenvectors (Matrix format):")
print(eigenvectors)

# 3. Access individual eigenvectors (use column slicing)
first_eigenvector = eigenvectors[:, 0]
second_eigenvector = eigenvectors[:, 1]

print("\nFirst eigenvector (corresponds to lambda = 3):", first_eigenvector)
```

```python
import numpy as np

# 8 points in a circle
points = np.array([
    [np.cos(i * np.pi / 4), np.sin(i * np.pi / 4)]
    for i in range(8)
])

# 1. Rotation: 30 degrees
a = np.radians(30)
R = np.array([
    [np.cos(a), -np.sin(a)],
    [np.sin(a),  np.cos(a)]
])

# 2. Scaling: [1.5, 0.8]
S = np.array([
    [1.5, 0],
    [0, 0.8]
])

# 3. Shearing: kx = 0.3
H = np.array([
    [1, 0.3],
    [0, 1]
])

# Compose transformations
M = H @ S @ R

# Apply to points
transformed = points @ M.T

print("Before transformation:")
print(points)

print("\nAfter transformation:")
print(transformed)

# Determinants
det_R = np.linalg.det(R)
det_S = np.linalg.det(S)
det_H = np.linalg.det(H)

det_M = np.linalg.det(M)
product = det_R * det_S * det_H

print("\nDeterminants:")
print("Rotation =", det_R)
print("Scaling  =", det_S)
print("Shearing =", det_H)

print("\nDeterminant of composed matrix =", det_M)
print("Product of individual determinants =", product)

print("Verified:", np.isclose(det_M, product))
```
