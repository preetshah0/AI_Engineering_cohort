1) Implement numerical_second_derivative(f, x) using numerical_derivative called twice. Verify that the second derivative of x^3 at x=2 is 12.

```python
def numerical_derivative(f, x, h=1e-5):
    return (f(x + h) - f(x)) / h


def second_derivative(f, x, h=1e-5):
    def first_derivative(t):
        return numerical_derivative(f, t, h)

    return numerical_derivative(first_derivative, x, h)


# Test with f(x) = x^3
f = lambda x: x**3

x = 2
print("Second derivative:", second_derivative(f, x))
```

2) Use gradient descent to find the minimum of f(x, y) = (x - 3)^2 + (y + 1)^2. Start from (0, 0). The answer should converge to (3, -1).

```python
def f(x, y):
    return (x - 3)**2 + (y + 1)**2


def gradient(x, y):
    dx = 2 * (x - 3)
    dy = 2 * (y + 1)
    return dx, dy


def gradient_descent(x, y, learning_rate=0.1, iterations=50):

    for i in range(iterations):

        # Calculate gradient
        dx, dy = gradient(x, y)

        # Update x and y
        x = x - learning_rate * dx
        y = y - learning_rate * dy

        if i % 5 == 0:
            print(f"Iteration {i}: x = {x:.4f}, y = {y:.4f}, f = {f(x,y):.4f}")

    return x, y


x, y = 0, 0

x, y = gradient_descent(x, y)

print("\nMinimum point:")
print("x =", x)
print("y =", y)
```

3) Add momentum to the gradient descent loop: maintain a velocity vector that accumulates past gradients. Compare convergence speed with and without momentum on f(x) = x^4 - 3x^2.

```python
def f(x):
    return x**4 - 3*x**2

def gradient(x):
    return 4*x**3 - 6*x

def gradient_descent(x, lr=0.01, iterations=1000):
    for i in range(iterations):
        x = x - lr * gradient(x)
    return x, iterations

def gradient_descent_momentum(x, lr=0.01, momentum=0.9, iterations=1000):
    velocity = 0

    for i in range(iterations):
        velocity = momentum * velocity - lr * gradient(x)
        x = x + velocity

    return x, iterations

x1, i1 = gradient_descent(2)
x2, i2 = gradient_descent_momentum(2)

print("Without momentum:", x1, f(x1), i1)
print("With momentum:", x2, f(x2), i2)
```
