1) Verify the inverse. Multiply A @ A.inverse_2x2() and confirm you get the identity matrix. Try it with three different 2x2 matrices. What happens when the determinant is zero?

```python
import numpy as np

def verify_inverse(A):
    # Calculate determinant
    det = np.linalg.det(A)
    print(f"Matrix A:\n{A}")
    print(f"Determinant: {det:.2f}")
    
    if np.isclose(det, 0):
        print("Determinant is zero! This matrix is singular and has no inverse.\n")
        return
        
    # Calculate inverse and multiply
    A_inv = np.linalg.inv(A)
    identity = A @ A_inv
    
    print(f"Inverse A^-1:\n{A_inv}")
    print(f"A @ A^-1 (Identity Matrix):\n{np.round(identity)}\n")

# 1. Three invertible matrices
matrix1 = np.array([[4, 7], [2, 6]])
matrix2 = np.array([[1, 2], [3, 4]])
matrix3 = np.array([[5, -2], [-3, 1]])

# 2. One matrix with a zero determinant
zero_det_matrix = np.array([[2, 4], [4, 8]])


verify_inverse(matrix1)
verify_inverse(matrix2)
verify_inverse(matrix3)
verify_inverse(zero_det_matrix)
```

2) Implement 3x3 inverse. Extend the Matrix class to compute inverses for 3x3 matrices using the adjugate method. Test it against NumPy's np.linalg.inv.

```python
import numpy as np

class Matrix3x3:
    def __init__(self, data):
        self.m = data

    def inverse(self):
        m = self.m
        
        # 1. Compute the 3x3 determinant
        det = (m[0][0] * (m[1][1] * m[2][2] - m[1][2] * m[2][1]) -
               m[0][1] * (m[1][0] * m[2][2] - m[1][2] * m[2][0]) +
               m[0][2] * (m[1][0] * m[2][1] - m[1][1] * m[2][0]))
        
        if det == 0:
            raise ValueError("Determinant is 0. Inverse does not exist.")

        # 2. Build the Adjugate Matrix 
        return [
            [
                (m[1][1]*m[2][2] - m[1][2]*m[2][1]) / det,
                (m[0][2]*m[2][1] - m[0][1]*m[2][2]) / det,
                (m[0][1]*m[1][2] - m[0][2]*m[1][1]) / det
            ],
            [
                (m[1][2]*m[2][0] - m[1][0]*m[2][2]) / det,
                (m[0][0]*m[2][2] - m[0][2]*m[2][0]) / det,
                (m[0][2]*m[1][0] - m[0][0]*m[1][2]) / det
            ],
            [
                (m[1][0]*m[2][1] - m[1][1]*m[2][0]) / det,
                (m[0][1]*m[2][0] - m[0][0]*m[2][1]) / det,
                (m[0][0]*m[1][1] - m[0][1]*m[1][0]) / det
            ]
        ]

A = [
    [1, 2, 3],
    [0, 1, 4],
    [5, 6, 0]
]

custom_inv = Matrix3x3(A).inverse()

numpy_inv = np.linalg.inv(A)

print("Match?", np.allclose(custom_inv, numpy_inv))
```
